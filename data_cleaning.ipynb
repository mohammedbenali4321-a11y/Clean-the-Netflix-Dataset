```python
import pandas as pd
data = pd.read_csv('netflix_titles.csv', sep=',', na_values=[''], quotechar='"')
data
```

```python
data.columns
```

```python
data.drop('cast',axis=1,inplace=True)
data.drop('description',axis=1,inplace=True)
```

```python
data.shape
```

```python
data.dropna(axis=0,how='any',inplace=True)
```

```python
data['type'].value_counts()
```

```python
data[["duration_value", "duration_type"]] = data["duration"].str.split(" ", expand=True)
data["duration_value"] = pd.to_numeric(data["duration_value"], errors="coerce")
data["duration_type"] = data["duration_type"].replace({
    "min": "Minutes",
    "Season": "Seasons",
    "Seasons": "Seasons"
})
```

```python
data.drop('duration',axis=1,inplace=True)
```

```python
data.head()
```

```python
data.drop('listed_in',axis=1,inplace=True)
```

```python
data.drop('title',axis=1,inplace=True)
```

```python
data['rating'].value_counts()
```

```python
data.rename(columns={'rating': 'content_rating'}, inplace=True)
```

```python
data['content_rating'] = data['content_rating'].replace({
    'TV-MA': 'Mature Adults',
    'TV-14': '14+ Teens',
    'R': 'Restricted',
    'TV-PG': 'Parental Guidance',
    'PG-13': 'Parents Strongly Cautioned',
    'PG': 'Parental Guidance',
    'TV-G': 'General Audience',
    'TV-Y7': 'Kids 7+',
    'TV-Y': 'Children',
    'NR': 'Not Rated',
    'G': 'General',
    'TV-Y7-FV': 'Kids Fantasy Violence',
    'UR': 'Unrated',
    'NC-17': 'Adults Only'
})
```

```python
data
```

```python
data['country'] = data['country'].str.split(',').str[0].str.strip()
```

```python
data['country'].value_counts()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.to_csv('final_dataset.csv')
```
