## Training Model by using ML

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('final_dataset.csv')
```

```python
df.head()
```

```python
df = df[['release_year', 'country']].dropna()

x = df[['release_year']]
y = df['country']

model = DecisionTreeClassifier()
model.fit(x, y)

pred = model.predict([[1993]])
pred
```
